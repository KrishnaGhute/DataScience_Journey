```python
%matplotlib inline

from autoviz.AutoViz_Class import AutoViz_Class
import matplotlib.pyplot as plt

AV = AutoViz_Class()

dft = AV.AutoViz(
    filename="titanic_train.csv",
    sep=",",
    depVar="",        # no target variable
    dfte=None,
    header=0,
    verbose=2,        # 🔥 IMPORTANT
    lowess=False,
    chart_format="png",  # 🔥 IMPORTANT
    max_rows_analyzed=150000,
    max_cols_analyzed=30,
)

plt.show()
```
