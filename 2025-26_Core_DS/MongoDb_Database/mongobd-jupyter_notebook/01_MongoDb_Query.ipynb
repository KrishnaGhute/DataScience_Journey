# Mongo Db with python - Quring Json Documents

```python
import pymongo
```

```python
client=pymongo.MongoClient('mongodb://127.0.0.1:27017/')
mydb=client['Employee']
empinfo=mydb.employeeinformation
```

```python
mydb
```

```python
records={
    'firstname':'Krishna',
    'lastname':'Ghute',
    'department':'DS',
    'qualification':'DS',
    'Age':19
}
empinfo.insert_one(records)
```

```python
records=[
    {
        "firstname": "Aarav",
        "lastname": "Sharma",
        "department": "DS",
        "qualification": "B.Tech",
        "Age": 21
    },
    {
        "firstname": "Isha",
        "lastname": "Verma",
        "department": "AI",
        "qualification": "M.Tech",
        "Age": 24
    },
    {
        "firstname": "Rohan",
        "lastname": "Patil",
        "department": "DS",
        "qualification": "B.Sc",
        "Age": 20
    },
    {
        "firstname": "Ananya",
        "lastname": "Deshmukh",
        "department": "ML",
        "qualification": "Ph.D",
        "Age": 28
    }
]
empinfo.insert_many(records)
```

```python
# Simple way of quering
empinfo.find_one()
# we are getting the first record
```

```python
# getting all the records
# Select*from employeeinformation
empinfo.find()
```

```python
# we are looping throug cursor
for records in empinfo.find():
    print(records)
```

```python
# Query the json documents on equality conditions
# select * from empinformation where firstname=Krishna

for records in empinfo.find({'firstname': 'Isha'}):
    print(records)
```

```python
# Query documents using query operators ($in,$lt,$gt)
for record in empinfo.find({'qualification':{'$in':['Ph.D','DS']}}):
    print(record)
```

```python
# And and Query operators
for record in empinfo.find({'qualification':'DS','Age':{'$lt':24}}):
    print(record)
```

```python
# OR operators
for record in empinfo.find({'$or':[{'firstname':'Krishna'},{'qualification':'B.tech'}]}):
    print(record)
```

```python
for record in empinfo.find({'$and':[{'firstname':'Krishna'},{'lastname':'Ghute'}]}):
    print(record)
```
