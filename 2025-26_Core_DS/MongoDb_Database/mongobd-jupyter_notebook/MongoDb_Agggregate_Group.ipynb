```python
from pymongo import MongoClient

# creation of object MongoClient
client=MongoClient()

# connecct with the portnumber and host
client=MongoClient('mongodb://localhost:27017/')

# Access database
mydatabase=client['Students']

# Access collection of the database
collection = mydatabase['studentscore']
data=[
      {"name": "Shyam", "subject": "Mathematics", "score": 88, "grade": "A"},
    {"name": "Ram", "subject": "Mathematics", "score": 92, "grade": "A+"},
    {"name": "Shyam", "subject": "Science", "score": 76, "grade": "B"},
    {"name": "Ram", "subject": "Science", "score": 84, "grade": "A"},
    {"name": "Shyam", "subject": "History", "score": 65, "grade": "C"},
    {"name": "Ram", "subject": "History", "score": 91, "grade": "A+"}
]

# collection.insert_many(data)
```

```python
# Find find ram and shyam total subjects
agg_result=collection.aggregate(
    [{
        '$group':
        {
            '_id':'$name',
            'Total subject records':{'$sum':1}
        }
    }]
)
for i in agg_result:
    print(i)
```

```python
# calculating the total score
agg_result=collection.aggregate(
    [{
        '$group':
        {
            '_id':'$name',
            'Total subject score':{'$sum':'$score'}
        }
    }]
)
for i in agg_result:
    print(i)
```

```python
# average socre based on the user
agg_result=collection.aggregate(
    [{
        '$group':
        {
            '_id':'$name',
            'stdavg':
            {
                '$avg':'$score'
            }
        }
    }]
)
for i in agg_result:
    print(i)
```

```python
# creating a new collection
data=[
    {"id": 1, "item": "Basmati Rice", "price": 450, "qty": 10, "date": datetime.datetime.utcnow()},
    {"id": 2, "item": "Masala Tea", "price": 120, "qty": 25, "date": datetime.datetime.utcnow()},
    {"id": 3, "item": "Turmeric Powder", "price": 85, "qty": 50, "date": datetime.datetime.utcnow()},
    {"id": 4, "item": "Moong Dal", "price": 110, "qty": 15, "date": datetime.datetime.utcnow()},
    {"id": 5, "item": "Ghee (500ml)", "price": 350, "qty": 8, "date": datetime.datetime.utcnow()},
    {"id": 6, "item": "Aloo Bhujia", "price": 40, "qty": 100, "date": datetime.datetime.utcnow()},
    {"id": 7, "item": "Coconut Oil", "price": 220, "qty": 12, "date": datetime.datetime.utcnow()}
]
```

```python
data
```

```python
mycollection=mydatabase['stores']
mycollection.insert_many(data)
```

```python
# claculating the average quantity and the average price
agg_result=mycollection.aggregate(
    [{
        '$group':
        {
            '_id':'$item',
            'avgamount':{'$avg':{'$multiply':['$price','$qty']}},
            'avgqty':{'$avg':'$qty'}
        }
    }]
)

for i in agg_result:
    print(i)
```

```python
# $project
for row in mycollection.aggregate([{'$project':{'item':1,'qty':1}}]):
    print(row)
```
