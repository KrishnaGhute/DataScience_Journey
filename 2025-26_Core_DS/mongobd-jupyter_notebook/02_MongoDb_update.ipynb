# MongoDb with python Updating Json Documents
## Functions to Discuss
- pymongo.collection.Collection.update_one()
- pymongo.collection.Collection.update_many()
- pymongo.collection.Collection.replace_one()

```python
import pymongo
client=pymongo.MongoClient('mongodb://127.0.0.1:27017/')
db = client['Employee']
inventry = db['inventry']
```

```python
inventry.update_one(
    {'item':'planner'},
    {'$set':{'size.uom':'m','status':'P'},
    '$currentDate':{'lastModifie':True}}
)
```

```python
inventry.update_many(
    {'qty':{'$lt':50}},
    {'$set':{'size.uom':'in','status':'PR'},
    '$currentDate':{'lastModifie':True}}
)
```

```python
inventry.replace_one(
    {'item':'paper'},
    {'item':'paper',
    'instock':[
    {'warehouse':'A','qty':60},
    {'warehouse':'B','qty':40}
    ]
    }
)
```
